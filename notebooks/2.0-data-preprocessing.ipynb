```python
# add time based column into df - hours of the day, day of the week, month, year, month year,
import pandas as pd

# Ensure 'DATE TIME' is in datetime format
df['DATE TIME'] = pd.to_datetime(df['DATE TIME'])

# Create time-based features
df['Hour'] = df['DATE TIME'].dt.hour               # Hour of the day (0-23)
df['DayOfWeek'] = df['DATE TIME'].dt.dayofweek      # Day of the week (0=Monday, 6=Sunday)
df['Month'] = df['DATE TIME'].dt.month              # Month (1-12)
df['Year'] = df['DATE TIME'].dt.year                # Year (e.g., 2020, 2021)
df['Month_Year'] = df['DATE TIME'].dt.to_period('M') # Month-Year (e.g., 2020-01)

# Display the updated DataFrame with new columns
print(df[['DATE TIME', 'Hour', 'DayOfWeek', 'Month', 'Year', 'Month_Year']].head())
```

```python
# Convert DATE TIME to datetime if it's not already, and sort by station and time
df['DATE TIME'] = pd.to_datetime(df['DATE TIME'])
df = df.sort_values(by=['STATION ID', 'DATE TIME'])

# Set the station and date as index (optional but helpful for time-based operations)
df.set_index(['STATION ID', 'DATE TIME'], inplace=True)

# Define the pollutants and lag intervals
pollutants = ['API', 'PM10 1H (µg/m3)', 'PM2.5 1H (µg/m3)']
lags = [1, 3, 6]

# Generate lagged features for each pollutant and each lag interval
for pollutant in pollutants:
    for lag in lags:
        df[f'{pollutant}_lag{lag}'] = df.groupby('STATION ID')[pollutant].shift(lag)

# Reset index if needed
df.reset_index(inplace=True)

# Display a sample of the DataFrame with new lagged features
print(df.head())
```

```python
# Fill NaN values in lagged columns
# Here, we replace NaN with 0 for lagged features (choose mean/median if preferable)
lagged_columns = [col for col in df.columns if 'lag' in col]
df[lagged_columns] = df[lagged_columns].fillna(0)

# Format API to have no decimal places
df['API'] = df['API'].round(0)
for lag in [1, 3, 6]:  # Ensure lagged API features also have no decimals
    df[f'API_lag{lag}'] = df[f'API_lag{lag}'].round(0)

# Format PM10 and PM2.5 columns to have up to 3 decimal places
df['PM10 1H (µg/m3)'] = df['PM10 1H (µg/m3)'].round(3)
df['PM2.5 1H (µg/m3)'] = df['PM2.5 1H (µg/m3)'].round(3)
for lag in [1, 3, 6]:  # Ensure lagged PM10 and PM2.5 features are also formatted
    df[f'PM10 1H (µg/m3)_lag{lag}'] = df[f'PM10 1H (µg/m3)_lag{lag}'].round(3)
    df[f'PM2.5 1H (µg/m3)_lag{lag}'] = df[f'PM2.5 1H (µg/m3)_lag{lag}'].round(3)

# Display the formatted DataFrame with no NaNs and adjusted decimal points
df.head()
```

```python
# Save as csv
df.to_csv('../data/interim/preprocessed_df.csv',index = False)
```
