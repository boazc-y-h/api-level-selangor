## Introduction & Background

In Malaysia, Selangor is an important economic hub chracterized by rapid urbanization and significant population density. The state includes diverse urban centers such as Banting, Batu Muda, Cheras, Klang, Kuala Selangor, Petaling Jaya, Putrajaya and Shah Alam. The state has numerous manufacturing facilities and industrial parks which attract both local and foreign investments. The rapid urbanization of Selangor has led to an influx of people seeking employment opportunities in these industries, resulting in one of the highest population densities in Malaysia. As more residents rely on personal and public transportation to commute to work, the volume of vehicles has surged. Consequently, the combination of high industrial output and dense traffic contributes to elevated levels of air pollution in the region.


Air pollution is a pressing global issue that significantly impacts both human health and the environment. Air pollution refers to the presence of harmful substances in the atmosphere such as gases, particulate matter (PM) and biological molecules, which can pose risks to human health, wildlife and the environment. Nowdays, these pollutants often originate from human activities including industrial emissions, vehicular exhaust and energy production. The consequences of air pollution extend beyond immediate health risks which contributing to long-term issues like respiratory diseases, cardiovascular conditions and even premature mortality.


This relationship underscores the critical importance of monitoring air quality indices like the Air Pollution Index (API) to safeguard public health. By tracking and analyzing air quality data, governments and organizations can implement effective policies and strategies to mitigate pollution and promote a healthier environment for all.


## Problem Statement

 According to the World Health Organization (WHO), air pollution is responsible for millions of premature deaths annually that contribute to respiratory diseases, cardiovascular conditions, and various other health problems. These deaths can be linked to increased prevalence of ambient molecular particles, such as nitrogen dioxide, carbon monoxide, ozone, and other particlulate matters (PM) (Anjum et al., 2024). Studies have demonstrated that elevated PM2.5 levels are directly linked to various health conditions. A recent cohort study found that for every 10 µg/m³ increase in PM2.5 exposure, there is a hazard ratio of 1.10 for non-accidental mortality and 1.09 for respiratory mortality, which show higher mortality risk upon exposure to PM (Qi et al., 2024).

In particular, Klang valley has been identified as one of the most polluted region in Selangor. In 2023, Batu Muda recorded PM2.5 readings that frequently fell within the highest 'moderate' pollution range. This can be attributed to high traffic volumes and industrial activities prevalent in the area. Batu Muda recorded 352 days of moderate API readings while the Cheras has 31 days of unhealthy API readings followed by Klang (19 days), as shown in Figure 1. # 2.0 Data Collection & Data Understanding

The dataset, provided by the Air Division of Jabatan Alam Sekitar (JAS), contains air quality readings from eight monitoring stations across Klang Valley. These stations include Banting, Klang, Kuala Selangor, Petaling Jaya, and Shah Alam in Selangor; Batu Muda and Cheras in the Federal Territory of Kuala Lumpur; and one station in Putrajaya, representing the Federal Territory of Putrajaya. The data was extracted from their database using an SQL query and shared in Excel format. It covers a date range from January 1, 2020, to **December 31, 2023**. Each year separated in different tab.

After merging all tabs, the dataset contains **280,495 entries** across 16 columns, capturing various environmental and air quality measurements at different stations. Key columns include STATION ID, LOCATION, and DATE TIME, which provide context for each measurement. The dataset contains air quality metrics, such as API (Air Pollution Index) and pollutant concentrations (e.g., PM10 1H (µg/m3), PM2.5 1H (µg/m3), SO2 1H (ppm), NO2 1H (ppm), O3 1H (ppm), and CO 1H (ppm)). Additionally, it includes meteorological variables like WIND DIRECTION 1H (°), WIND SPEED 1H (m/s), RELATIVE HUMIDITY 1H (%), and Ambient Temperature 1H (°c). 

The data types include float64 for numerical values, int64 for YEAR, and object for categorical information such as STATION ID and LOCATION. Some columns have missing values, specifically SO2 1H (ppm), NO2 1H (ppm), O3 1H (ppm), and CO 1H (ppm), with around **245,440 non-null entries**, indicating gaps in pollutant measurements. The dataset requires careful handling of missing values and potentially some data type conversions for further analysis.
