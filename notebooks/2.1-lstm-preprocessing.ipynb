# Data Preprocessing for LSTM Model Training

Preprocessing steps for LSTM Models:

1. Exclusion of Lag Features:

LSTM models inherently learn temporal dependencies from sequential data. Therefore, manually adding lag values is unnecessary and may even introduce redundancy. The model architecture itself captures patterns across previous time steps.

2. Creation of Station-Specific Datasets:

Time series forecasting requires data to maintain a continuous and chronologically ordered sequence. For environmental forecasting like API, each monitoring station’s readings should be treated separately unless an aggregated metric (e.g., average API) is used. Mixing data from different stations without accounting for their spatial and temporal context can confuse the model and degrade performance.

## 1.0 Remove Redundant Features

```python
#import df from drive
import pandas as pd
df = pd.read_csv('../data/raw/api-data-jas.csv')
df.head()

# Identifying columns in dataset
df.columns.to_list()
```

```python
# Drop redundant columns and rearrange target to the end
df_copy = df.drop(['LOCATION','API','YEAR','DATE_TIME'], axis=1)
df_copy['API'] = df['API']
```

## 2.0 Create Station-Specific Datasets

```python
# List stations present in dataset
stations = df['STATION ID'].unique().tolist()
```

```python
# Prepare LSTM dataset for each station
for station in stations:
    df_station = df_copy[df_copy['STATION ID']==station].drop('STATION ID', axis=1)
    df_station.dropna(axis=1, inplace=True)
    print(station + str(df_station.shape))
    df_station.to_csv("../data/raw/"+station+".csv", index=False)
```
